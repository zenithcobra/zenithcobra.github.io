```python
import requests
from bs4 import BeautifulSoup
import json

def scrape_matchups_to_json(url):
    """
    Scrapes the team matchups section from the given URL and converts it to JSON format.

    Args:
        url (str): The URL of the webpage to scrape.

    Returns:
        dict: A dictionary containing the scraped matchups data.
    """
    # Send a GET request to the webpage
    response = requests.get(url)
    response.raise_for_status()  # Raise an exception for HTTP errors

    # Parse the webpage content with BeautifulSoup
    soup = BeautifulSoup(response.text, 'html.parser')

    # Find the matchups section
    matchups = []
    matchup_divs = soup.find_all('div', class_='matchup')  # Adjust class name based on the webpage structure

    for div in matchup_divs:
        # Extract team names, winning chances, and other details
        teams = div.find_all('img', alt=True)
        team1 = teams[0]['alt']
        team2 = teams[1]['alt']

        winning_chances = div.find_all('div', class_='winning-chance')  # Adjust class name
        chance1 = winning_chances[0].text.strip()
        chance2 = winning_chances[1].text.strip()

        starters = div.find_all('div', class_='starter')  # Adjust class name
        starter1 = starters[0].text.strip()
        starter2 = starters[1].text.strip()

        sources = div.find_all('div', class_='source')  # Adjust class name
        source1 = sources[0].text.strip()
        source2 = sources[1].text.strip()

        matchups.append({
            "team1": team1,
            "team2": team2,
            "chance1": chance1,
            "chance2": chance2,
            "starter1": starter1,
            "starter2": starter2,
            "source1": source1,
            "source2": source2
        })

    # Convert the matchups data to JSON format
    matchups_json = json.dumps(matchups, indent=4)
    return matchups_json


# Example usage
url = "https://moneypuck.com/"
matchups_json = scrape_matchups_to_json(url)
print(matchups_json)
```

```python
from requests_html import HTMLSession

def scrape_matchups_with_requests_html(url):
    """
    Scrapes the team matchups section from the given URL using requests-html.

    Args:
        url (str): The URL of the webpage to scrape.

    Returns:
        dict: A dictionary containing the scraped matchups data.
    """
    session = HTMLSession()
    response = session.get(url)
    response.html.render()  # Render JavaScript

    # Parse the rendered HTML
    matchups = []
    matchup_divs = response.html.find('.matchup')  # Adjust selector based on the webpage structure

    for div in matchup_divs:
        team1 = div.find('.team1', first=True).text
        team2 = div.find('.team2', first=True).text
        chance1 = div.find('.chance1', first=True).text
        chance2 = div.find('.chance2', first=True).text
        starter1 = div.find('.starter1', first=True).text
        starter2 = div.find('.starter2', first=True).text
        source1 = div.find('.source1', first=True).text
        source2 = div.find('.source2', first=True).text

        matchups.append({
            "team1": team1,
            "team2": team2,
            "chance1": chance1,
            "chance2": chance2,
            "starter1": starter1,
            "starter2": starter2,
            "source1": source1,
            "source2": source2
        })

    return matchups


# Example usage
url = "https://moneypuck.com/"
matchups = scrape_matchups_with_requests_html(url)
print(matchups)
```

```python
def convert_string_to_number_list(input_string):
    """
    Converts a string of numbers separated by '-' into a list of integers.

    Args:
        input_string (str): The input string (e.g., '3-3-2-4-3-2-3-7-4').

    Returns:
        list: A list of integers (e.g., [3, 3, 2, 4, 3, 2, 3, 7, 4]).
    """
    return [int(x) for x in input_string.split('-') if x.isdigit()]





from itertools import product

def generate_sequences_and_find_distribution(current_sequence, target_distribution, lookahead=4):
    """
    Generates all possible sequences for the next `lookahead` spots, calculates the distribution
    of values (1, 2, or 3), and finds sequences that result in the target distribution.

    Args:
        current_sequence (list): The current sequence (e.g., [0, 0, 1, 0, 0, 0, 1, 0, 0, 0]).
        target_distribution (float): The target distribution percentage (e.g., 35%).
        lookahead (int): The number of spots to look ahead (default is 3).

    Returns:
        list: A list of sequences that result in the target distribution.
    """
    # current_distribution
    current_value = 0
    for y in current_sequence:
        if y > 0:
            current_value += 1
    current_distribution = round((current_value / len(current_sequence)), 2)

    # Generate all possible combinations for the next `lookahead` spots
    possible_values = [0, 1,]
    possible_combinations = list(product(possible_values, repeat=lookahead))
    # print(possible_combinations)

    # Find sequences that match the target distribution
    matching_sequences = []
    for combination in possible_combinations:
        # Create the new sequence
        new_sequence = current_sequence + list(combination)
        size = len(new_sequence)
        value = 0
        for x in new_sequence:
            if x > 0:
                value += 1

        percent = (value / size)
        print(current_sequence, current_distribution,'%', list(combination), round(percent, 2),'%')


    return matching_sequences

# Example usage
input_string = '0-0-1-0-0-0-1-0-0-0'
number_list = convert_string_to_number_list(input_string)
current_sequence = number_list[::-1]
target_distribution = 35.0
current_distribution = 2/10
print(f"Current sequence: {current_sequence} current distribution is {current_distribution} and target distribution is {target_distribution}%")
matching_sequences = generate_sequences_and_find_distribution(current_sequence, target_distribution)

print(f"Matching sequences with {target_distribution}% distribution:")
for seq in matching_sequences:
    print(seq)
```
