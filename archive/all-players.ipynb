```python
import csv

# Read the CSV file and extract the required columns
input_file = 'text_output/war.csv'
output_file = 'all_players_with_teams.csv'

# Initialize lists
players_with_teams = []

# Read the input file
with open(input_file, mode='r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    header = next(reader)  # Skip the header row
    for row in reader:
        common_name = row[0]  # First column
        team_id = row[5]      # Sixth column
        players_with_teams.append((common_name, team_id))

# Remove duplicates
unique_players_with_teams = list(set(players_with_teams))

# Write to the output file
with open(output_file, mode='w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['common_name', 'team_id'])  # Write header
    writer.writerows(unique_players_with_teams)
```

```python
# Open the CSV file and extract the second column
teams = set()  # Use a set to automatically remove duplicates
with open(output_file, mode='r', newline='', encoding='utf-8') as csvfile:
    reader = csv.reader(csvfile)
    next(reader)  # Skip the header row
    for row in reader:
        teams.add(row[1])  # Add the second column (team_id) to the set

# Sort the teams alphabetically
sorted_teams = sorted(teams)

# Save the sorted list to a new file
with open('all_teams.csv', mode='w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['team_id'])  # Write header
    for team in sorted_teams:
        writer.writerow([team])
```

```python
# Read all_players_with_teams.csv as a list of lists
with open('all_players_with_teams.csv', mode='r', newline='', encoding='utf-8') as file:
    all_players_with_teams = list(csv.reader(file)) # team_id[1]

# Read team_ref.csv as a list of lists
with open('team_ref.csv', mode='r', newline='', encoding='utf-8') as file:
    team_ref_data = list(csv.reader(file)) #team_id[1] # full name [2]

for x in all_players_with_teams:
    for y in team_ref_data:
        if x[1] == y[1]:
            x.append(y[2])  # Append the full name to the player data
            break

# Write the updated all_players_with_teams to a new CSV file
with open('all_players_with_teams_full_name.csv', mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(all_players_with_teams)  # Write the updated data

```

```python
# Open the input file and read its contents
input_file = 'text_output/all_players_with_teams.csv'
output_file = 'text_output/all_players_with_teams_switched.csv'

with open(input_file, mode='r', newline='', encoding='utf-8') as infile:
    reader = csv.reader(infile)
    rows = list(reader)

# Switch columns 2 and 3 for each row
for row in rows[1:]:  # Skip the header row
    row[1], row[2] = row[2], row[1]

# Save the updated rows to a new file
with open(output_file, mode='w', newline='', encoding='utf-8') as outfile:
    writer = csv.writer(outfile)
    writer.writerows(rows)
```
