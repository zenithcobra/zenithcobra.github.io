```python

import os
import json
from datetime import datetime, timedelta


def save_to_text(content, supplied_filename):
    """
    Saves today's matches to a text file in the 'data' folder. If the file already exists, 
    it archives the existing file with yesterday's date in the 'data/archived_data' folder.

    Args:
        string of text: The data to save to the txt file.
        supplied_filename (str): The base name of the file to save (e.g., 'todays_matches').

    Returns:
        None
    """
    # Directories
    data_dir = "data"
    archived_dir = os.path.join(data_dir, "archived_data")
    os.makedirs(data_dir, exist_ok=True)  # Ensure the 'data' directory exists
    os.makedirs(archived_dir, exist_ok=True)  # Ensure the 'archived_data' directory exists

    # File paths
    today_date = datetime.now().strftime("%Y-%m-%d")
    yesterday_date = (datetime.now() - timedelta(days=1)).strftime("%Y-%m-%d")
    file_path = os.path.join(data_dir, f"{supplied_filename}")
    archived_file_path = os.path.join(archived_dir, f"{supplied_filename}_{yesterday_date}.txt")

    # Check if the file already exists in the 'data' folder
    if os.path.exists(file_path):
        # Archive the existing file with yesterday's date
        if not os.path.exists(archived_file_path):
            os.rename(file_path, archived_file_path)
            print(f"Archived existing file to {archived_file_path}")
        else:
            print(f"Archived file already exists: {archived_file_path}")

    # Save today's matches to the JSON file in the 'data' folder
    # with open(file_path, "w") as json_file:
    #     json.dump(list_of_dicts, json_file, indent=4)
    with open(file_path, "w") as file:
        file.write(content)

    print(f"Today's data saved to {file_path}")


def read_json_file(file_path):
    """
    Reads a JSON file from the specified file path, ensuring the file exists.

    Args:
        file_path (str): The path to the JSON file to be read.

    Returns:
        dict or list: The parsed JSON data from the file.

    Raises:
        FileNotFoundError: If the specified file does not exist.
        ValueError: If the file is not a valid JSON file.
        Exception: For any other errors encountered while reading the file.
    """
    # Check if the file exists
    if not os.path.exists(file_path):
        raise FileNotFoundError(f"The file at path '{file_path}' does not exist.")
    
    try:
        # Open and read the JSON file
        with open(file_path, 'r') as file:
            data = json.load(file)
        return data
    except json.JSONDecodeError as e:
        raise ValueError(f"The file at path '{file_path}' is not a valid JSON file: {e}")
    except Exception as e:
        raise Exception(f"An error occurred while reading the file: {e}")
    
# convert to html table
# Function to generate an HTML table from the sample data
def generate_html_table(data):
    # Define the table headers
    headers = ["batter",
"batter_id",
"batter_team",
"pitcher",
"pitcher_id",
"pitcher_team",
"pitcher_team_id",
"AB",
"H",
"HR",
"AVG",
"RBI",
"OBP",
"OPS",
"team_id",
"all_games_played",
"all_H",
"all_HR",
"all_RBI",
"all_HRpg",
"all_fHRpg",
"all_Hpg",
"all_fHpg",
"all_RBIpg",
"all_fRBIpg",
"all_HR24",
"all_HR24pg",
"all_fHR24pg",
"all_HR_record",
"all_H_record",
"all_RBI_record"]
    
    # Start the HTML table
    html = "<table border='1'>\n"
    
    # Add the table headers
    html += "  <tr>\n"
    for header in headers:
        html += f"    <th>{header}</th>\n"
    html += "  </tr>\n"
    
    # Add the table rows
    for row in data:
        html += "  <tr>\n"
        for header in headers:
            # Get the value for the current header, handle 'null' and missing keys
            value = row.get(header, " ")
            if value is None:
                value = " "
            html += f"    <td>{value}</td>\n"
        html += "  </tr>\n"
    
    # Close the HTML table
    html += "</table>"
    
    return html

sample_data = read_json_file('data/ace_bvp_with_streaks.json')
# print(sample_data)
for z in sample_data:
    print(z.get('batter'))
#     for x, y in z.items():
#         print(x)
# Generate the HTML table
html_table = generate_html_table(sample_data)

# Print the HTML table
# print(html_table)
save_to_text(html_table, "bvp.html")
```
