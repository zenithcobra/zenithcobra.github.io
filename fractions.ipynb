```python
allowed_denominators = [4, 8, 16, 32, 64, 128]

def closest_fractions(x, denominators, top_n=5):
    results = []

    for d in denominators:
        n = round(x * d)
        value = n / d
        diff = abs(value - x)

        results.append((f"{n}/{d}", value, diff))

    results.sort(key=lambda r: r[2])
    return results[:top_n]

x = 0.100

for frac, value, diff in closest_fractions(x, allowed_denominators):
    print(f"{frac:>8} = {value:.6f}  diff={diff:.6f}")






```

```python
from fractions import Fraction

def closest_fractions(x, max_den=128, top_n=5):
    candidates = []

    for d in range(1, max_den + 1):
        n = round(x * d)
        frac = Fraction(n, d)
        diff = abs(float(frac) - x)
        candidates.append((str(frac), diff))

    candidates.sort(key=lambda t: t[1])
    return candidates[:top_n]

print(closest_fractions(0.100))
```
