```python
import statsapi
from datetime import datetime

# get the proper formatted date
mlb_date = datetime.now().strftime("%m/%d/%Y")

# how to get the last 10 games for a team
sched = statsapi.schedule(start_date='01/01/2025',end_date=mlb_date,team=141)
newlist = sorted(sched, key = lambda k: k["game_date"], reverse=True)
for a in newlist[0:10]:
    beans = statsapi.get("game", {"gamePk": a.get('game_id')})

    persons = []
    for x,y in beans.get('liveData').get('boxscore').get('teams').get('away').get('players').items():
        persons.append(x)

    for x in persons:
        p_id = beans.get('liveData').get('boxscore').get('teams').get('away').get('players').get(x).get('person').get('id')
        name = beans.get('liveData').get('boxscore').get('teams').get('away').get('players').get(x).get('person').get('fullName')
        position = beans.get('liveData').get('boxscore').get('teams').get('away').get('players').get(x).get('position').get('type')
        hits = beans.get('liveData').get('boxscore').get('teams').get('away').get('players').get(x).get('stats').get('batting').get('hits')
        # Fetch season stats
        stats = statsapi.player_stat_data(p_id, 'hitting', type="season", season=2025).get('stats')

        # Check if stats is a list and process it
        if isinstance(stats, list) and stats:  # Ensure stats is a non-empty list
            for stat_entry in stats:
                fstats = stat_entry.get('stats')
                hits = fstats.get('hits')
                print(f"{name} -> {hits} hits")
                # for key, value in fstats.items():
                #     print(f"{key}: {value}")
# for x,y in beans.get('gameData').get('players').items():
#     print(x,y)

# Game_data_keys
# copyright
# gamePk
# link
# metaData
# gameData
# liveData
```
