```python


import os
import json
from bs4 import BeautifulSoup
import statsapi
from datetime import datetime, timedelta
from fractions import Fraction

# -------

# File path for the JSON file
json_file_path = "text_output/teams_last_10_games.json"

# Ensure the file exists
if not os.path.exists(json_file_path):
    raise FileNotFoundError(f"{json_file_path} does not exist.")

# Read the JSON file into a dictionary
with open(json_file_path, "r", encoding="utf-8") as file:
    teams_last_10_games = json.load(file)

json_file_path2 = "text_output/teams_playing_today.json"
# Ensure the file exists
if not os.path.exists(json_file_path2):
    raise FileNotFoundError(f"{json_file_path2} does not exist.")

# Read the JSON file into a dictionary
with open(json_file_path2, "r", encoding="utf-8") as file:
    teams_playing_today = json.load(file)

csv_file_path = "text_output/all_all_teams_with_ids.csv"
# Ensure the file exists
if not os.path.exists(csv_file_path):
    raise FileNotFoundError(f"{csv_file_path} does not exist.")

# Read the csv file into a list of lists
import csv
with open(csv_file_path, "r", encoding="utf-8") as file:
    csv_reader = csv.reader(file)
    all_teams_with_ids = [row for row in csv_reader]

teams_with_ids_list = []
for y in teams_playing_today:
    for x in all_teams_with_ids:
        if y == x[3]:
            teams_with_ids_list.append([x[0],y])

# remove duplicates
teams_with_ids_list = list(set(tuple(x) for x in teams_with_ids_list))
# Convert back to list of lists
teams_with_ids_list = [list(x) for x in teams_with_ids_list]

schedule1 = statsapi.schedule(game_id='777687')
# print(schedule1[0].get('winning_team'))
print(schedule1)


# for a in teams_with_ids_list:
#     print(a[1])
#     for x in teams_last_10_games.get(a[0]):
#         print(x)

```

```python
import csv

# File path
csv_file_path = 'text_output/team_ref.csv'

# Read the CSV file into a list of lists
teams = []
with open(csv_file_path, mode='r', encoding='utf-8') as file:
    reader = csv.reader(file)
    teams = list(reader)  # Convert the reader object to a list of lists

# Print the list of lists
# for row in teams:
#     print(row[2])


# open and read the file 'text_output/team_ref.csv' into a list


# for x in teams:
#     print(x[3])


new_teams_list = []
for x in teams:
    team_data = statsapi.lookup_team(x[2])
    # print(team_data)
    if team_data:
        team_id = team_data[0]['id']
        team_list = [team_id,x[0],x[1],x[2],x[3],x[4]]
        new_teams_list.append(team_list)

# save new_teams_list to a csv file named 'text_output/all_teams_with_ids.csv'
import csv
with open('text_output/all_all_teams_with_ids.csv', mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['team_name', 'team_id'])  # Write header
    writer.writerows(new_teams_list)  # Write data
```
