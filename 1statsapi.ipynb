```python
# Roster and processing logic

import statsapi

roster_string = statsapi.roster(143)
roster_list_unprocessed = roster_string.split('\n')
roster_list = []
for x in roster_list_unprocessed:
    roster_line = ''
    for y in x.split(' ')[4:]:
        roster_line = roster_line + y + ' '
    roster_list.append(roster_line.lstrip())

for a in roster_list:
    print(a)
    
```
