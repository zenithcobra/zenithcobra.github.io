```python
import mlbstatsapi
mlb = mlbstatsapi.Mlb()

import statsapi
# Open the file and read all the names into a list
with open('text_output/all_players.txt', 'r') as file:
    all_players = [line.strip() for line in file]

# Extract only the names (last two words)
player_names = [" ".join(line.split()[-2:]) for line in all_players]

# Display the list of names
# player_names

new_all_players = []
for x in player_names:
    player_info = statsapi.lookup_player(x, gameType="R", season=None, sportId=1)
    beans = ''
    # print(x)
    team_id = player_info[0].get('currentTeam').get('id')
    team = mlb.get_team(team_id)
    # print(team.name)
    new_line_string = f'{x},{team.name}'
    new_all_players.append(new_line_string)

# Write the updated list to a new file
with open('text_output/all_players_with_teams.csv', 'w') as file:
    for player in new_all_players:
        file.write(player + '\n')
```
