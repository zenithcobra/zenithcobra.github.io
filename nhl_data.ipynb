```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV file
data = pd.read_csv("NHL_data/nhl_skaters_2025_latest.csv")

# Focus on the column of interest
shots_on_goal = data["I_F_shotsOnGoal"]

# Correlation matrix to find relationships
correlation_matrix = data.corr()
print(correlation_matrix["I_F_shotsOnGoal"].sort_values(ascending=False))

# Visualize relationships
sns.pairplot(data, vars=["I_F_shotsOnGoal", "I_F_xGoals", "I_F_shotAttempts"])
plt.show()

# Scatter plot for specific relationships
plt.scatter(data["I_F_xGoals"], data["I_F_shotsOnGoal"])
plt.xlabel("I_F_xGoals")
plt.ylabel("I_F_shotsOnGoal")
plt.title("Relationship between xGoals and Shots on Goal")
plt.show()
```
