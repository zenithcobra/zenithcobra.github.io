```python
def string_to_binary_list(string):
    """
    Converts a string of 'W' and 'L' into a list of 1's and 0's.

    Args:
        string (str): The input string containing 'W' and 'L' separated by '-'.

    Returns:
        list: A list of 1's for 'W' and 0's for 'L'.
    """
    return [1 if char == 'W' else 0 for char in string.split('-') if char]

def analyze_binary_list(binary_list):
    """
    Analyzes a list of 1's and 0's to calculate its length and count the number of 1's and 0's.

    Args:
        binary_list (list): A list of 1's and 0's.

    Returns:
        dict: A dictionary containing the length of the list, the count of 1's, and the count of 0's.
    """
    length = len(binary_list)
    count_ones = binary_list.count(1)
    count_zeros = binary_list.count(0)
    return {
        "length": length,
        "count_ones": count_ones,
        "count_zeros": count_zeros
    }

# Example usage
# string = 'W-W-L-W-W-W-L-L-L-W-W-W-W-W-L-L-L-L-W-W-W-W-L-L-L-L-W-L-L-L-L-L-W-W-L-W-W-L-W-L-W-L-W-L-W-L-W-L-L-W-W-L-W-L-L-L-L-W-W-L-L-W-W-W-W-W-L-W-W-L-L-L-W-L-W-W-W-L-W-L-L-L-W-L-L-W-W-W-L-L-W-W-W-L-W-W-L-W-W-L-L-L-L-W-L-W-L-W-W-L-W-L-W-L-L-L-'
string = 'W-W-L-W-W-W-L-L-L-W-W-W-W-W-L-L-L-L-W-W-W-W-L-L-L-L-W-L-L-L-L-L-W-W-L-W-W-L-W-L-W-L-W-L-W-L-W-L-L-W-W-L-W-L-L-L-L-W-W-L-L-W-W-W-W-W-L-W-W-L-L-L-W-L-W-W-W-L-W-L-L-L-W-L-L-W-W-W-L-L-W-W-W-L-W-W-L-W-W-L-L-L-L-W-L-W-L-W-W-L-W-L-W-L-L-L-'
string2 = 'W-L-L-W-W-L-W-L-W-W-W-L-W-L-L-W-L-W-W-W-W-W-W-L-L-W-W-L-W-L-L-L-W-L-L-L-L-L-L-L-L-L-W-L-W-L-L-L-L-L-W-W-L-W-L-L-L-L-W-L-L-L-W-L-W-W-L-L-W-L-L-L-W-L-W-L-L-L-W-L-W-L-W-L-L-W-L-L-W-W-W-L-L-W-L-L-W-W-W-L-L-L-L-W-L-L-L-W-W-L-W-L-W-L'


binary_list = string_to_binary_list(string2)
analysis = analyze_binary_list(binary_list)

print("Binary List:", binary_list)
print("Analysis:", analysis)
```
